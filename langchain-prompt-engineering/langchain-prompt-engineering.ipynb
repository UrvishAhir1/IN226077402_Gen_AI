# LangChain Prompt Engineering: Building a Mini Prompt Engine

## Introduction

In traditional prompt engineering, prompts are often hardcoded using string formatting, which makes them difficult to reuse and maintain. 

This notebook demonstrates how to design a structured and reusable prompt system using LangChain. Instead of writing static prompts, we use PromptTemplate and ChatPromptTemplate to dynamically generate prompts based on user inputs.

## Objectives

- Replace hardcoded prompts with reusable templates
- Build multi-input prompt systems
- Create different prompt styles (teaching, interview, storytelling)
- Implement chat-based prompting using roles
- Add input validation for robustness
- Design a mini prompt generation pipeline

## Workflow

User Input > Validation > Template Selection > Prompt Generation  Output

```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

---

## Replace Hardcoded Prompt

In this task, we convert a static prompt into a reusable PromptTemplate using LangChain.

```python
template_1 = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

output_1 = template_1.format(topic="Artificial Intelligence")
print(output_1)
```

---

## Multi-Input Prompt System

This template takes multiple inputs such as topic, audience, and tone.

```python
template_2 = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

print(template_2.format(topic="AI", audience="beginners", tone="friendly"))
print(template_2.format(topic="Python", audience="kids", tone="fun"))
print(template_2.format(topic="Deep Learning", audience="engineers", tone="technical"))
```

---

## Prompt Variations Engine

Different templates are created for different styles of prompting.

```python
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 interview questions about {topic}"
)

story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

topic = "Machine Learning"

print(teaching_template.format(topic=topic))
print(interview_template.format(topic=topic))
print(story_template.format(topic=topic))
```

---

## Chat-Based Prompt System

Using ChatPromptTemplate to simulate role-based prompting.

```python
def get_chat_prompt(role, topic):
    
    if role == "teacher":
        system_msg = "You are a helpful teacher."
    elif role == "interviewer":
        system_msg = "You are an interviewer."
    elif role == "motivator":
        system_msg = "You are a motivational speaker."
    else:
        system_msg = "You are a helpful assistant."

    chat_template = ChatPromptTemplate.from_messages([
        ("system", system_msg),
        ("user", "Explain {topic}")
    ])

    return chat_template.format_messages(topic=topic)


messages = get_chat_prompt("teacher", "Neural Networks")
for msg in messages:
    print(f"{msg.type}: {msg.content}")
```

---

## Input Validation

Validates user inputs before generating prompts.

```python
def validate_inputs(audience, tone):
    
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        audience = "beginner"

    if tone not in valid_tone:
        tone = "formal"

    return audience, tone
```

---

## Prompt Generator

Combines validation, template selection, and prompt generation.

```python
def generate_prompt(topic, audience, tone, style):
    
    audience, tone = validate_inputs(audience, tone)

    if style == "teaching":
        template = "Explain {topic} clearly step by step for {audience} in a {tone} tone"
    elif style == "interview":
        template = "Ask 3 interview questions about {topic} for {audience} in a {tone} tone"
    elif style == "storytelling":
        template = "Explain {topic} as a story for {audience} in a {tone} tone"
    else:
        template = "Explain {topic} for {audience} in a {tone} tone"

    prompt = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template=template
    )

    return prompt.format(topic=topic, audience=audience, tone=tone)


print(generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))
```

---

## Template Reusability Test

Using a single template with multiple inputs.

```python
reuse_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms"
)

topics = ["AI", "Python", "Machine Learning", "Deep Learning", "Data Science"]

for t in topics:
    print(reuse_template.format(topic=t))
```

---

## Prompt Playground

This section demonstrates how the prompt engine behaves with different inputs.

It helps validate that the system is reusable and can handle multiple scenarios.

```python
test_cases = [
    ("AI", "beginner", "fun", "teaching"),
    ("Python", "intermediate", "casual", "storytelling"),
    ("Machine Learning", "expert", "formal", "interview"),
    ("Data Science", "beginner", "fun", "storytelling"),
    ("Neural Networks", "expert", "formal", "teaching")
]

for topic, audience, tone, style in test_cases:
    print("INPUT:")
    print(f"Topic: {topic}, Audience: {audience}, Tone: {tone}, Style: {style}")
    
    output = generate_prompt(topic, audience, tone, style)
    
    print("OUTPUT:")
    print(output)
    print("-" * 60)
```
